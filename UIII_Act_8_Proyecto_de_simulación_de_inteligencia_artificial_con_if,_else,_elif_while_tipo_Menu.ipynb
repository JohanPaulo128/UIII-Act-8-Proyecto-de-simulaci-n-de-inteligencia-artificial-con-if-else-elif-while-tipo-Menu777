### Escritor y libro

```python
print("--- Asistente Creativo IA ---")
opcion = ""
while opcion != "3":
    print("\nMenú:\n1. Sugerir Género\n2. Idea de Trama\n3. Salir")
    opcion = input("Elige una opción: ")

    if opcion == "1":
        animo = input("¿Cómo te sientes hoy? (feliz/triste): ")
        if animo == "feliz":
            print("IA: ¡Escribe una Comedia Romántica!")
        elif animo == "triste":
            print("IA: Tu género es el Drama Profundo.")
        else:
            print("IA: Suena a que debes escribir Ciencia Ficción.")
    elif opcion == "2":
        lugar = input("¿Prefieres el bosque o la ciudad?: ")
        if lugar == "bosque":
            print("IA: Un mapa perdido aparece bajo un árbol viejo.")
        else:
            print("IA: Un robot encuentra un libro prohibido en el metro.")
    elif opcion == "3":
        print("¡Suerte con tu libro!")
    else:
        print("Opción no válida, intenta de nuevo.")
```
