```python
print("--- BIENVENIDO A ESCRITOR-IA v1.0 ---")
print("Tu mentor creativo automatizado")

while True:
    print("\n*******************************")
    print("       MENÚ DE OPCIONES")
    print("*******************************")
    print("1. Sugerir un Género")
    print("2. Generar Idea de Trama")
    print("3. Consultar frase inspiradora")
    print("4. SALIR")
    print("*******************************")

    opcion = input("Elige una opción (1-4): ")

    if opcion == "1":
        print("\n--- SECCIÓN: SUGERENCIA DE GÉNERO ---")
        animo = input("¿Cómo está tu energía hoy? (alta/baja): ")
        if animo == "alta":
            print("IA: Deberías escribir una Novela de Aventuras o Acción.")
        elif animo == "baja":
            print("IA: Es un buen momento para un Drama o Poesía.")
        else:
            print("IA: No reconozco ese estado, mejor escribe Suspenso.")

    elif opcion == "2":
        print("\n--- SECCIÓN: GENERADOR DE TRAMAS ---")
        lugar = input("¿Prefieres el pasado o el futuro?: ")
        if lugar == "pasado":
            print("IA: Un caballero encuentra una brújula que apunta al espacio.")
        elif lugar == "futuro":
            print("IA: Un hacker descubre que vive dentro de una pintura antigua.")
        else:
            print("IA: Elige una opción válida para darte una idea.")

    elif opcion == "3":
        print("\n--- SECCIÓN: INSPIRACIÓN ---")
        print("IA: 'Escribir es la manera más profunda de leer la vida'.")

    elif opcion == "4":
        print("\n¡Gracias por crear conmigo! Guardando manuscritos... Adiós.")
        break

    else:
        print("\n[!] Opción no válida. Por favor, marca 1, 2, 3 o 4.")
```
